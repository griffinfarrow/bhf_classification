Notebook comparing the performance of different efficient net models 

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import pickle
```

```python
with open('b0/train_metrics.txt', 'rb') as file: 
    b0_train_data = pickle.load(file)

with open('b0/val_metrics.txt', 'rb') as file: 
    b0_val_data = pickle.load(file)

with open('b2/train_metrics.txt', 'rb') as file: 
    b2_train_data = pickle.load(file)

with open('b2/val_metrics.txt', 'rb') as file: 
    b2_val_data = pickle.load(file)

with open('b4/train_metrics.txt', 'rb') as file: 
    b4_train_data = pickle.load(file)

with open('b4/val_metrics.txt', 'rb') as file: 
    b4_val_data = pickle.load(file)

with open('resnet_18/train_metrics.txt', 'rb') as file: 
    r18_train_data = pickle.load(file)

with open('resnet_18/val_metrics.txt', 'rb') as file: 
    r18_val_data = pickle.load(file)  
```

```python
epochs = [i for i in range(1, 11)]
```

```python
fig, ax = plt.subplots()
ax.plot(epochs, [val['exact_match'] for val in b0_train_data], linestyle = '--', marker = 'x', label = 'training, b0')
ax.plot(epochs, [val['exact_match'] for val in b0_val_data], linestyle = '-', marker = 'x', label = 'validation, b0')

ax.plot(epochs, [val['exact_match'] for val in b2_train_data], linestyle = '--', marker = 'x', label = 'training, b2')
ax.plot(epochs, [val['exact_match'] for val in b2_val_data], linestyle = '-', marker = 'x', label = 'validation, b2')

ax.plot(epochs, [val['exact_match'] for val in b4_train_data], linestyle = '--', marker = 'x', label = 'training, b4')
ax.plot(epochs, [val['exact_match'] for val in b4_val_data], linestyle = '-', marker = 'x', label = 'validation, b4')

ax.plot(epochs, [val['exact_match'] for val in r18_train_data], linestyle = '--', marker = 'x', label = 'training, r18')
ax.plot(epochs, [val['exact_match'] for val in r18_val_data], linestyle = '-', marker = 'x', label = 'validation, r18')

ax.set_xlabel(r"Epoch")
ax.set_ylabel(r"Accuracy")
ax.set_ylim(0.35, 0.5)
ax.legend()
fig.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(2, 1, sharex=True)
ax[0].plot(epochs, [np.mean(res['per_class_f1']) for res in b0_train_data], linestyle = '-', marker = 'x', label = 'b0')
ax[1].plot(epochs, [np.mean(res['per_class_f1']) for res in b0_val_data], linestyle = '-', marker = 'x', label = 'b0')

ax[0].plot(epochs, [np.mean(res['per_class_f1']) for res in b2_train_data], linestyle = '-', marker = 'x', label = 'b2')
ax[1].plot(epochs, [np.mean(res['per_class_f1']) for res in b2_val_data], linestyle = '-', marker = 'x', label = 'b2')

ax[0].plot(epochs, [np.mean(res['per_class_f1']) for res in b4_train_data], linestyle = '-', marker = 'x', label = 'b4')
ax[1].plot(epochs, [np.mean(res['per_class_f1']) for res in b4_val_data], linestyle = '-', marker = 'x', label = 'b4')

ax[0].plot(epochs, [np.mean(res['per_class_f1']) for res in r18_train_data], linestyle = '-', marker = 'x', label = 'r18')
ax[1].plot(epochs, [np.mean(res['per_class_f1']) for res in r18_val_data], linestyle = '-', marker = 'x', label = 'r18')

ax[1].set_xlabel(r"Epoch")
ax[0].set_ylabel(r"Train: Avg F1 Score")
ax[1].set_ylabel(r"Test: Avg F1 Score")
ax[0].legend()
ax[1].legend()
fig.tight_layout()
plt.show()
```
