```python
import h5py
import numpy as np 
import matplotlib.pyplot as plt
import os
```

```python
root_path = os.path.abspath("../")
data_path = os.path.join(root_path, "data")
filepath = os.path.join(data_path, "proc_images.h5")

def read_preprocessed_images(filepath):
        
    with h5py.File(filepath, 'r') as f:
        raw_names = f["img_names"][:]
        images = f["images"][:]
    image_names = [n.decode("utf-8") for n in raw_names]
    image_set = dict(zip(image_names, images))
    
    return image_set
```

```python
image_set = read_preprocessed_images(filepath)
```

```python
plt.figure()
img = image_set["train_000003.png"]
plt.imshow(img, cmap="gray")
plt.show()
```
