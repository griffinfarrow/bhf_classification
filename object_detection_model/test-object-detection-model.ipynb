```python
import torch
import torchvision
import torchvision.transforms as T
import numpy as np
import cv2
from PIL import Image
import matplotlib.pyplot as plt
```

```python
def predict(image_path, model, out_image_path='/kaggle/working/processed_image.png', threshold=0.8):
    model.eval()
    image = Image.open(image_path).convert("RGB")
    transform = T.Compose([T.ToTensor()])
    img_tensor = transform(image).unsqueeze(0).to(device)

    with torch.no_grad():
        predictions = model(img_tensor)

    boxes = predictions[0]['boxes'].cpu().numpy()
    scores = predictions[0]['scores'].cpu().numpy()

    # Read image with OpenCV
    image_cv = cv2.imread(image_path)

    for i, box in enumerate(boxes):
        if scores[i] > threshold:  # Confidence threshold
            x1, y1, x2, y2 = map(int, box)
            cv2.rectangle(image_cv, (x1, y1), (x2, y2), (0, 255, 0), 3)

    cv2.imwrite(out_image_path, image_cv)

# Run prediction
predict("/kaggle/input/test-dataset/train_package/train_000047.png", model)
```
